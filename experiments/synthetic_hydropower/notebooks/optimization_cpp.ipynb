# GIVP C++ × balanço hídrico sintético

O binário C++ chama o GIVP header-only e mantém um worker Python com pipes bidirecionais.

```c++17
#include <cstdlib>
#include <string>
const auto* root = std::getenv("GIVP_ROOT");
if (root == nullptr) throw std::runtime_error("set GIVP_ROOT");
std::string base(root);
std::string build = "cmake -S \"" + base + "/experiments/synthetic_hydropower/clients/cpp\" -B \"" + base + "/experiments/synthetic_hydropower/output/cpp-build\" && cmake --build \"" + base + "/experiments/synthetic_hydropower/output/cpp-build\"";
if (std::system(build.c_str()) != 0) throw std::runtime_error("C++ build failed");
std::string run = "\"" + base + "/experiments/synthetic_hydropower/output/cpp-build/synthetic_hydropower_optimize\" \"" + base + "/experiments/synthetic_hydropower/interop/v1/optimization_definition.json\"";
if (std::system(run.c_str()) != 0) throw std::runtime_error("C++ demo failed");
```
