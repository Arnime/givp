# GIVP Rust × balanço hídrico sintético

O binário experimental mantém uma conexão JSON Lines com a referência Python.

```rust
let root = std::env::var("GIVP_ROOT").expect("set GIVP_ROOT");
let manifest = format!("{}/experiments/synthetic_hydropower/clients/rust/Cargo.toml", root);
let definition = format!("{}/experiments/synthetic_hydropower/interop/v1/optimization_definition.json", root);
let output = std::process::Command::new("cargo").args(["run", "--manifest-path", &manifest, "--bin", "synthetic-hydropower-optimize", "--", &definition]).output().expect("run Rust demo");
println!("{}", String::from_utf8_lossy(&output.stdout));
assert!(output.status.success());
```
